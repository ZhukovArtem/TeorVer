# Правило сложения и умножения

Многие комбинаторные задачи могут быть решены с помощью двух правил - правила умножения и правила сложения

## Правило суммы
Если некоторый объект А может быть выбран из совокупности объектов m споссобами, а другой объект B может быть выбран n способами, то выбрать либо А, либо B можно m + n способами

## Правило произведения
Если объект можно выбрать из совокупности объектов m способами и после каждого такого выбора объект B можно выбрать т способами, то пара объектов в указанном порядке может быть выбрана m * n способами

Сколько комбинаций можно составить переставляя буквы в слове "гипербола"?

```python
from math import factorial

factorial(9)
```

Сколько комбинаций можно составить из слова "колобок"?

```python
from math import factorial

factorial(7) / (factorial(3) * factorial(2))
```

Размещение без повторений - это выборка из n элементов k элементов без повторений и с учетом порядка следования элементов

Размещение с повторением - это выборка из n элементов k элементов без повторения и без учета порядка следования элементов

Сколькими способами можно выбрать из 13 человек старосту и дежурного

```python
from math import factorial

factorial(13) / factorial(13 - 2)
```

1.67. Сколько раз нужно провести испытание, чтобы с вероятностью, не меньшей Р, можно было утверждать, что по крайней
мере один раз произойдет событие, вероятность которого в каждом
испытании равна р? Дать ответ при р = 0,4 и Р = 0,8704. 

Ai - благоприятный исход в i испытании

p[Ai] = 0.4. A - событие произошло хотя бы раз. -А - событие не произошло ни разу

-А = -А1 * -А2 * .. * -Аn 

P(-A) = P(-A1) * P(-A2) * .. * P(-An) = 0.6 ** n

P(A) = 1 - P(-A) = 1 - 0.6 ** n

P(A) >= 0.8704 \n
0.6 ** n <= 0.1896 \n
0.6 ** n <= 0.6 ** 4 \n
n <= 4
