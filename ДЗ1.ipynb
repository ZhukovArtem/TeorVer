```python
n = 4
n_ = factorial(n)
m = 0

for i in range(0, n + 1):
    m = m + ((-1) ** i / factorial(i))
m = m * factorial(n)
p = m / n_

print(f'{p:.4f}')
```

**Пример 1.74**
>Два стрелка сделали по одному выстрелу в мишень. Вероятность попадания в мишень для первого стрелка равна 0,6, а для второго - 0,3. В мишени оказалась одна пробоина. Найти вероятность того, что она принадлежит первому стрелку. 

```python
p1 = 0.6
q1 = 1 - p1
p2 = 0.3
q2 = 1 - p2

# Условные вероятности одного попадания равны одному только при попадании либо только первого, либо только второго стрелка
# Остальные случаи опустим
# По формуле Байеса можно вывести:
p = p1 * q2 / (p1 * q2 + q1 * p2)
round(p, 4)
```
